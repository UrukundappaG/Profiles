# Encapsulation

There are three types of Access Modifiers
1. Public
2. Protected
3. Private

**Public**
The least restrictive modifier, allowing access to members from anywhere in the program. Public variables are not prefixed with any underscores.

**Protected**
Allows members to be accessed by members within a class or by its subclasses. Protected variables are prefixed with a single underscore ( _ ).

**Private**
The most restrictive modifier, limiting access to members within the same class. Private variables are prefixed with a double underscore ( __ ).

```python
class Student:
    def __init__(self,name,stu_id,mobile):
        self.name=name       # public access modifier
        self._stu_id=stu_id  # protected
        self.__mobile=mobile   #private


    def info(self):
        return f'Your name: {self.name}\nstu_id:{self._stu_id}\nMobile: {self.__mobile}'
```

# Inheritance
- To inherit the properties of the Parent class

1. Parent Class   / Base class
2. Child Class    / Derived class

```python
class Student:     # Parent
    def __init__(self,name,stu_id,mobile):
        self.name=name       
        self._stu_id=stu_id 
        self.__mobile=mobile 


    def info(self):
        return f'Your name: {self.name}\nstu_id:{self._stu_id}\nMobile: {self.__mobile}'
```

```python
class Result(Student):
    def __init__(self,name,stu_id,mobile,score):
        super().__init__(name,stu_id,mobile)  # inheriting from parent class
        self.score=score

    def result(self):
        print(f'Name: {self.name}\
                \nstu_id: {self._stu_id}\
              \nScore: {self.score}\
              \nThe score has been sent on {self.__mobile}')
    
```

```python
r=Result('Rani','ABCD1234F','9876543210','95%')
```

```python
r.result()
```

# Abstraction

- ABC  =  Abstract Base Classes 
- abstractmethod

```python
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod   # decorator
    def pay(self, amount):
        pass
    # @abstractmethod
    # def pin(self,pip):
    #     pass

    def balance(self):
        print('Hello memmber')

    
```

```python
class CreditCardPayment(Payment):
    def pay(self, amount):
        print(f"Paid ${amount} using Credit Card.")
    def pin(self,pin):
        print(pin)
```

```python
ccp=CreditCardPayment()

ccp.pay(100)
```
