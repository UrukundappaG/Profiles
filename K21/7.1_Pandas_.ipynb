# Dataframe.iloc[ ] method
- used when the index label of a data frame is something other than numeric series of 0, 1, 2, 3….n
- or in case the user doesn’t know the index label.

#### pandas.DataFrame.iloc[]

### 1: Extracting single row and comparing with .loc[]

### 1: Extracting single row and comparing with .loc[]

- loc   ==> LABEL
- iloc  ==> INDEX

```python
bond = pd.read_csv("jamesbond.csv", index_col= "Film")
bond.head(3)

print(bond.loc['Dr. No'] )

#iloc --> Index
bond.iloc[0]

row1 = bond.iloc[25]
row1
```

### 2: Extracting multiple rows with index

```python
### 2: Extracting multiple rows with index
bond.iloc[ [4, 5, 6, 7]  ]
```

### 3: Slicing

```python
### 3: Slicing

bond.iloc[4:8]

row1 = bond.iloc[ [5, 6, 7]  ]
row2 = bond.iloc[5:8]
row1 == row2
```

# Filtering

```python
import pandas as pd
data = [['Sourabh',32, 98000], ['Disha', 26, 118000], ['Pankaj', 45, 76000]]

df = pd.DataFrame(data,
                  index =["Choclate", "Pizza", "Pasta"],
                  columns=['Name','Age',"salary"])
df

#Pandas filter() Columns by Labels
df.filter(items=['Name', 'salary'])

# select columns by regular expression
df.filter(regex='e$', axis=1)
df.filter(regex='y$', axis=1)
```

##DateTime Objects in Pandas
###Convert string to DateTime

Pandas has a built-in function called to_datetime()that converts date and time in string format to a DateTime object.



```python
import datetime as dt
```

```python
df = pd.DataFrame({'date': ['04/03/2022', '05/03/2022', '06/03/2022'],
                   'patients': [16, 19, 11]})

df.info()
```

```python
df.dtypes
```

```python

df['date'] = pd.to_datetime(df['date'])
df.info()
```

```python
print(type(df['date']))
```

##Convert integer to DateTime

```python
df0 = pd.DataFrame({'time_of_death': ['2020-7-13 21:15:0',
                            '2020-7-9 17:30:30',
                            '2020-12-10 4:45:12'],
                   'patientID': [101, 23, 48]})

df0['time_of_death'] = pd.to_datetime(df0['time_of_death'],
                                      format="%Y-%m-%d %H:%M:%S")
df0
```
