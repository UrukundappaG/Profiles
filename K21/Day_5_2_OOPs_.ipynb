P1 :- WAP that uses class to store the name and marks of students.


```python
class Students:
   def __init__(self, name ):
       self.name = name   #Initialize variable
       self.marks = []

   def enterMarks(self):
       for i in range(2):
           m = int(input("ENTER MARKS OF %s IN SUBJECT %d : " %(self.name, i+1)))
           self.marks.append(m)

   def display(self):
       print(self.name , "got", self.marks)

s1 = Students("Sourabh")
s1.enterMarks()
s1.display()
```

P3:- WAP that has a class Circle. Use a class variable to define the value to  constant PI. Use this a variable to calculate area and circumference of a circle with specified radius.



```python
class Circle:
   PI = 3.14159
   def __init__(self, radius):
       self.radius = radius
   def area(self):
       return Circle.PI*self.radius*self.radius
   def circumference(self):
       return 2*Circle.PI*self.radius

C = Circle(7.5)
print("AREA =", C.area() )
print("CIRCUMFRENCE =", C.circumference() )
```
