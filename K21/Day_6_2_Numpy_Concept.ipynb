# 2. Array Stacking and Splitting
---
## 2.1Stacking



Several arrays can be stacked together along different axes.

- **np.vstack:** To stack arrays along vertical axis.

- **np.hstack:** To stack arrays along horizontal axis.

- **np.concatenate:** To stack arrays along specified axis (axis is passed as argument).

```python
a = np.array([[1,3],
             [4,7]])

b = np.array([[7,9],
             [3,1]])
```

## 2.2. Splitting

- Array splitting can be vertical, horizontal, or depth-wise.

For splitting, we have these fuctions:

- **np.hsplit:** Split array along horizontal axis.

- **np.vsplit:** Split array along vertical axis.

- **np.array_split:** Split array along specified axis.

```python
a = np.array([[1,3,5,7,8,9,10, 12],
              [1,3,5,7,8,9,10,56]])
```

# 3. Array Broadcasting
---


#Broadcasting

- Broadcasting is an operation of matching the dimensions of differently shaped arrays in order to be able to perform further operations on those arrays (eg per-element arithmetic).

- A very powerful feature


image.png

```python
a = np.array([2,4,5])
b = np.array([2,2,2])

a * b
```

```python
a = np.array([2,4,5])
b = np.array([2])

a * b
```
