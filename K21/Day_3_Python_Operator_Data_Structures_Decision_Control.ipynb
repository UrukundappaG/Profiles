## Math Module
- math.pi: Represents the mathematical constant π (pi).
- math.e: Represents the mathematical constant e.
- math.sqrt(x): Computes the square root of x.
- math.pow(x, y): Computes x raised to the power of y.
- math.exp(x): Computes e raised to the power of x.
- math.ceil(x): Rounds x to the smallest integer greater than or equal to x.
- math.floor(x): Rounds x to the largest integer less than or equal to x.
- math.factorial(x): Computes the factorial of x.
- math.trunc(x): Truncates x to an integer.
- math.fabs(x): Returns the absolute value of x.

```python
import math

print("π:", math.pi)
print("e:", math.e)
print("Square root of 25:", math.sqrt(25))
print("2 raised to the power of 3:", math.pow(2, 3))
#print("Sine of 30 degrees:", math.sin(math.radians(30)))
print("Ceil of 3.7:", math.ceil(3.7))
print("Floor of 3.7:", math.floor(3.7))
print("Factorial of 5:", math.factorial(5))
#print("GCD of 12 and 18:", math.gcd(12, 18))
#print("Approximately equal:", math.isclose(1.23, 1.24, rel_tol=0.01))
```
