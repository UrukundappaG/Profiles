```python
# Design an basic calculator + - / *


# 3 inputs
# enter fist number
# enter the second number
# Enter the operation you want to perform ---->   + - / *

# Invalid operator

10
20
+ or add

30

10
20
- or sub

-10
```

```python
num_1 = eval(input('Enter the first number : '))
num_2 = eval(input('Enter the second number : '))
op = input('Enter the operation you want to perform : ')


if (op == '+') or (op == 'add'):
    print(num_1+ num_2)
elif (op == '-') or (op == 'sub'):
    print(num_1-num_2)
elif (op == '/') or (op == 'div'):
    print(num_1/num_2)
elif (op == '*') or (op == 'mul'):
    print(num_1*num_2)
else:
    print('Please enter the valid operator')
```
