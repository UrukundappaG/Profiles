```python
def max( numbers ):

  max = numbers[0]  # [2,5555,3,6,7,188,12,89]

  for i in range(1,8):
    if numbers[i] > max: # max = 89
      max =  numbers[i]
  return max

```



Write a program(WAP) to calculate volume of cuboid using default arguments.



```python
def volume(l, w=3, h=4):
    print("LENGTH :",l, "\t WIDTH :", w, "\t HEIGHT :", h)
    return l*w*h

print ("VOLUME :", volume(5,4,2))
print ("VOLUME :", volume(5,4))
print ("VOLUME :", volume(5))
```

```python
def name(firstName, lastName):
    separator = ' '
    n = firstName + separator + lastName
    return n

print(name('Sourabh' , "Taneja"))
```

```python

def is_Prime(num):
  if num < 2:
      return False
  i = 2
  while i ** 2 <= num:
      if num % i == 0:
          return False
      i += 1
  return True


primes = 0
composites = 0

while True:
    try:
        num = int(input())
        if num == -1 :
            break
        elif num < 2:  # Numbers less than 2 are neither prime nor composite
            continue
        elif is_Prime(num):
            primes += 1
        else
            composites += 1
            
    except ValueError:
        print("Invalid input. Please enter an integer.")

print(f"\n Number of prime numbers entered: {primes}")
print(f"Number of composite numbers entered: {composites}")
```
