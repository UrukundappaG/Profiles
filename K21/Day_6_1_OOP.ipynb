## **Protected Members**
---

A protected member is intended to be used only inside the class or by its subclasses.



**Using Protected Attributes**

```python
class Base:
    def __init__(self):
        self._protected_var = "I am protected"

class Derived(Base):
    def show(self):
        print(self._protected_var)

d = Derived()
d.show()
```

## **Private Members**
---
---

- A private member is strictly for internal use inside the class.

- Prefix it with two underscores __.



```python
class Employee:
    def __init__(self):
        self.__salary = 50000

    def display_salary(self):
        print("Salary is:", self.__salary)

emp = Employee()
emp.display_salary()
print(emp.__salary)
```

Inheritance means that one class (child) can use the properties and methods of another class (parent).

image.png

### Syntax

image.png

```python
class Animal:
    def speak(self):
        print("Animal speaks")

class Dog(Animal):
    def bark(self):
        print("Dog barks")


d = Dog()
d.speak()
d.bark()
```

 ### Single Inheritance

```python
class A:
    def show_A(self):
        print("This is A")

class B(A):  # B inherits from A
    def show_B(self):
        print("This is B")

b = B()
b.show_A()
b.show_B()
```

## Multiple Inheritance

```python
class A:
    def feature_A(self):
        print("Feature A")

class B:
    def feature_B(self):
        print("Feature B")

class C(A, B):  # C inherits from both A and B
    def feature_C(self):
        print("Feature C")

c = C()
c.feature_A()
c.feature_B()
c.feature_C()
```

## Multilevel Inheritance

```python
class Grandparent:
    def show_gp(self):
        print("Grandparent")

class Parent(Grandparent):
    def show_p(self):
        print("Parent")

class Child(Parent):
    def show_c(self):
        print("Child")

c = Child()
c.show_gp()
c.show_p()
c.show_c()
```

## Hierarchical Inheritance

```python
class Parent:
    def show_parent(self):
        print("Parent")

class Child1(Parent):
    def show_c1(self):
        print("Child 1")

class Child2(Parent):
    def show_c2(self):
        print("Child 2")

c1 = Child1()
c2 = Child2()

c1.show_parent()
c2.show_parent()
```

The word polymorphism means:

  **"Many Forms"**
  - Poly = many,
  - morph = forms


Polymorphism means the same function or method name behaves differently depending on the object or context.

```python
class Animal:
    def speak(self):
        print("Animal speaks")

class Dog(Animal):
    def speak(self):
        print("Dog barks")

class Cat(Animal):
    def speak(self):
        print("Cat meows")

# Polymorphic behavior
for animal in [Dog(), Cat()]:
    animal.speak()
```

```python
class Duck:
    def talk(self):
        print("Quack quack")

class Human:
    def talk(self):
        print("Hello!")

def call_talk(obj):
    obj.talk()  # Works for both, no need for common parent class

call_talk(Duck())   # Quack quack
call_talk(Human())  # Hello!
```

```python
def add(x, y):
    return x + y

print(add(2, 3))         # Numbers → Output: 5
print(add("Hi ", "Sam")) # Strings → Output: Hi Sam
print(add([1], [2, 3]))  # Lists → Output: [1, 2, 3]
```
